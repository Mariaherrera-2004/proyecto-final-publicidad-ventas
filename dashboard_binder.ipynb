# Dashboard — Publicidad digital y ventas en e-commerce

Este notebook lanza la aplicación Dash (`app.py`) en segundo plano y la muestra automáticamente en el entorno de Binder.

**Instrucciones:** En el menú superior haga clic en **Run → Run All Cells**. Espere 5 segundos y verá el dashboard interactivo desplegado abajo.

```python
import subprocess, time, os

# 1. Liberar el puerto 8050 si habia una sesion previa
subprocess.run(["pkill", "-9", "-f", "app.py"], stderr=subprocess.DEVNULL)
time.sleep(1)

# 2. Iniciar el servidor del dashboard en segundo plano sin warnings en pantalla
proceso = subprocess.Popen(
    ["python", "app.py"],
    cwd=os.getcwd(),
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)
time.sleep(5)
print("Dashboard iniciado exitosamente (PID:", proceso.pid, ")")
```

```python
import os
from IPython.display import HTML, display, IFrame

# Obtener la ruta base del proxy en Binder
prefix = os.environ.get('JUPYTERHUB_SERVICE_PREFIX', '/')
proxy_url = f"{prefix}proxy/8050/"

# Enlace para abrir en pestana completa si se prefiere
display(HTML(f'''
    <div style="padding: 12px; background-color: #f1f5f9; border: 1px solid #cbd5e1; border-radius: 8px; margin-bottom: 15px;">
        <span style="font-size: 1.05em; font-weight: 600; color: #1e293b;">El dashboard esta listo:</span> 
        <a href="{proxy_url}" target="_blank" style="font-size: 1.1em; font-weight: bold; color: #4338ca; text-decoration: underline; margin-left: 8px;">
            Abrir dashboard en pantalla completa (nueva pestana)
        </a>
    </div>
'''))

# Incrustar el dashboard interactivo directamente en el cuaderno
display(IFrame(src=proxy_url, width="100%", height="800"))
```
