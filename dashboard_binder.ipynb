# Dashboard — Publicidad digital y ventas en e-commerce

Este notebook lanza `app.py` (la aplicación Dash) como proceso en segundo plano y expone el enlace para verla dentro de Binder, usando `jupyter-server-proxy` (ya incluido en `requirements.txt`).

**Pasos:** Cell → Run All. Espere ~5 segundos y verá el dashboard interactivo desplegado abajo.

```python
import subprocess, time, os

# Lanza la app Dash en segundo plano en el puerto 8050
proceso = subprocess.Popen(["python", "app.py"], cwd=os.getcwd())
time.sleep(5)
print("Dashboard iniciado en segundo plano (PID:", proceso.pid, ")")
```

```python
import os
from IPython.display import HTML, display, IFrame

# Prefijo dinámico de la sesión en Binder / JupyterHub
prefix = os.environ.get('JUPYTERHUB_SERVICE_PREFIX', '/')
proxy_url = f"{prefix}proxy/8050/"

display(HTML(f'''
    <div style="padding:12px; background-color:#e0e7ff; border-radius:8px; margin-bottom:15px;">
        <a href="{proxy_url}" target="_blank" style="font-size:1.2em; font-weight:bold; color:#4f46e5;">
            👉 Abrir el dashboard interactivo en una pestaña nueva
        </a>
    </div>
'''))

# Incrustar el dashboard directamente en el cuaderno
display(IFrame(src=proxy_url, width="100%", height="750"))
```

Si el dashboard o el enlace no cargan de inmediato, espere unos segundos y ejecute la celda nuevamente.
